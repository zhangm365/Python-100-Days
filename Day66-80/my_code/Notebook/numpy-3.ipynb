```python
# 数组运算

import numpy as np

## 1. 数组与标量的运算
arr1 = np.arange(1, 10)
print(arr1.shape)

print(arr1 + 10)
print(arr1 * 10)

## 2. 数组与数组的运算
arr2 = np.array([1, 1, 1, 2, 2, 2, 3, 3, 3])
print(arr1 + arr2)
print(arr1 * arr2)
print(arr1 ** arr2)
```

```python
## 通用一元函数
print(np.sqrt(arr1))
print(np.log2(arr1))
```

```python
## 通用二元函数:两个数组
arr3 = np.array([[4, 5, 6], [7, 8, 9]])
arr4 = np.array([[1, 2, 3], [3, 2, 1]])

print(np.maximum(arr3, arr4))
print(np.power(arr3, arr4))

## 点积
arr5 = np.array([[1, 2], [2, 3], [3, 4]])
print(np.dot(arr3, arr5))
```

```python
## 广播机制
arr5 = np.array([[0, 0, 0], [1, 1, 1], [2, 2, 2], [3, 3, 3]])
arr6 = np.array([1, 2, 3])
print(arr5 + arr6)

arr7 = np.array([[1], [2], [3], [4]])
print(arr5 + arr7)

arr8 = np.array([[1], [2], [3]])
print(arr8.shape)

arr9 = np.array([[1, 2, 3]])
print(arr9.shape)

print(arr8 + arr9)
```
