```python
# DataFrame：每个列的数据类型不需要相同

import pandas as pd
import numpy as np

## 通过二维数组创建 DataFrame
scores = np.random.randint(60, 101, (5, 3))
courses = ['Python', 'Java', 'C++']
stu_ids = np.arange(1001, 1006)
df1 = pd.DataFrame(data=scores, index=stu_ids, columns=courses)
df1

## 通过字典创建 DataFrame
scores = {
    '语文': [78, 72, 93, 88, 93],
    '数学': [95, 98, 86, 96, 87],
    '英语': [80, 75, 82, 93, 82],
}
stu_ids = np.arange(1001, 1006)
df2 = pd.DataFrame(data=scores, index=stu_ids)
df2
```
