```python
"""
numpy：Python 科学计算库
"""
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# 使用 array 函数，通过 list 创建数据对象
arr1 = np.array([1, 2, 3, 4, 5])
arr1

arr2 = np.array([[1, 2, 3], [4, 5, 6]])
arr2


# arange 函数：指定取值范围和跨度创建数组对象，左闭右开
arr3 = np.arange(0, 20, 2)
arr3

arr4 = np.linspace(-1, 1, 11)
arr4

# 数组的索引计算
arr5 = np.arange(1, 10)
arr5

print(arr5[0], arr5[arr5.size - 1])
print(arr5[-arr5.size], arr5[-1])

arr6 = np.array([[1, 2, 3], [10, 11, 12], [7, 8, 9]])
arr6
```
