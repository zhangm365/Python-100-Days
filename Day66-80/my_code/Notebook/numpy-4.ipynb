```python
# 1. 向量：矢量，同时具有大小和方向。
import numpy as np

u = np.array([5, 1, 3])
m1 = np.array([4, 5, 1])
m2 = np.array([5, 1, 5])

## 1.1. 向量的点积: 两个向量对应分量求和，结果是一个标量。几何意义是两个向量的模乘以二者夹角的余弦。
vec_dot = np.dot(u, m1)    # 通过 dot 函数进行计算
## 向量的模长通过 linalg.norm 函数进行计算
vec_norm = np.linalg.norm(u) * np.linalg.norm(m1)
print(f'向量夹角的余弦值：{vec_dot / vec_norm}')

vec_dot = np.dot(u, m2)
## 1.2. 向量的模长通过 linalg.norm 函数进行计算
vec_norm = np.linalg.norm(u) * np.linalg.norm(m2)
print(f'向量夹角的余弦值：{vec_dot / vec_norm}')

# 2. 行列式：n 阶方阵。
```
