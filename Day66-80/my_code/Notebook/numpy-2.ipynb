```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

arr1 = np.random.randint(1, 100, 10)
arr1

# 计算总和，均值和中位数
print('数组的总和为：')
print(arr1.sum())
print(np.sum(arr1))

print('数组的算术平均值为：')
print(arr1.mean())
print(np.mean(arr1))

print('数组的中位数为：')
print(np.median(arr1))

print('数组的分位数为：')
print(np.quantile(arr1, 0.5))    # 第二个参数为 0.5 表示计算 50% 分位数
```

```python
# 极值、全距和四分位距离
print(arr1)

print('数组中的最大值：')
print(arr1.max())
print(np.amax(arr1))

print('数组中的最小值：')
print(arr1.min())
print(np.amin(arr1))

print('数组的全距：')
print(np.ptp(arr1))

print('数组的四分位距离：')
q1, q3 = np.quantile(arr1, [0.25, 0.75])
print(q3 - q1)
```

```python
# 方差，标准差和变异系数
print('方差为：')
print(arr1.var())
print(np.var(arr1))

print('标准差为：')
print(arr1.std())
print(np.std(arr1))

print('变异系数：')
print(arr1.std() / arr1.mean())
```

```python
# 箱线图
plt.boxplot(arr1, showmeans = True)
plt.ylim([-20, 120])
plt.show()
```
