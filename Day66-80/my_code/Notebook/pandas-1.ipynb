```python
"""
pandas 是一套强大的分析结构化数据的工具集，提供了专门用于数据分析的类型、方法和函数。

核心数据类型：Series 和 DataFrame
"""

import pandas as pd
import numpy as np

ser1 = pd.Series(data = [120, 380, 300, 600], index = ['一季度', '二季度', '三季度', '四季度'])
ser1

## Series 对象的运算
ser1 += 10

## 通过字典创建 Series 对象
ser2 = pd.Series({'一季度': 320, '二季度': 180, '三季度': 400, '四季度': 500})
ser2

## 索引运算
ser1.iloc[1]    # 整数索引
ser1['二季度']    # 自定义索引

## 切片索引
ser2[1:3]

ser2['二季度':'四季度']    # 使用自定义索引时，结束索引对应的元素也可以取到。

## Series 对象的属性

print(ser2.dtype)
print(ser2.index)
print(ser2.values)
print(ser2.hasnans)  # 是否有缺失值
```

```python
## 统计
print(ser2.count())  # 非缺失值的数量
print(ser2.sum())    # 求和
print(ser2.mean())   # 求平均值
print(ser2.median())  # 求中位数
print(ser2.std())    # 求标准差
print(ser2.var())    # 求方差
```

```python
# Series 对象的 apply() 和 map() 方法
ser6 = pd.Series(['cat', 'dog', np.nan, 'rabbit'])
ser6

ser6.map({'cat': 'kitten', 'dog': 'puppy'})

ser6. map('I am a {}'.format, na_action='ignore')  # na_action='ignore' 忽略缺失值

ser7 = pd.Series([10, 20 ,30], index = ['London', 'Paris', 'Berlin'])
ser7

ser7.apply(np.square)

ser7.apply(lambda x, value: x - value, args=(5,))  # lambda 函数:第一个参数是数据系列中的数据，第二个参数是传入的值
```

```python
# 排序
ser8 = pd.Series([35, 96, 12, 57, 25, 89], index=['grape', 'apple', 'banana', 'orange', 'kiwi', 'pear'])
ser8.sort_values()  # 按值升序排序

ser8.sort_index()  # 按索引升序排序

ser8.nlargest(3)  # 返回最大的三个值
ser8.nsmallest(2)  # 返回最小的两个值
```

```python
# 绘制图表
import matplotlib.pyplot as plt

ser9 = pd.Series({'Q1': 400, 'Q2': 520, 'Q3': 300, 'Q4': 600})

# ser9.plot(kind='bar')

ser9.plot(kind='pie', autopct='%.1f%%', pctdistance=0.65)
plt.show()  # autopct 用于显示百分比
```
